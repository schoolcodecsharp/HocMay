## Chất lượng
Outlier theo IQR là cờ kiểm tra, không tự động xóa dòng.

```python
print(train.isna().sum())
print('Duplicates:', train.duplicated().sum())
print('Target cap:', (train[TARGET] >= 5.0).sum())
q1, q3 = train.quantile(.25), train.quantile(.75)
iqr = q3 - q1
((train < q1 - 1.5 * iqr) | (train > q3 + 1.5 * iqr)).sum()
```

```python
import matplotlib.pyplot as plt
train.hist(figsize=(14, 10), bins=40)
plt.tight_layout()
plt.show()
```
